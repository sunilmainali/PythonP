```python
with open("poem.txt") as f:
    c=f.read()
    if ("twinkle" in c):
        print('twinkle is present')
    else:
        print('twinkle is not present')
```

```python
def generateTables(n):
    table=""
    for i in range(1,11):
         table += f"{n} X {i} = {n*i}\n" 
    with open(f"tables/table_{n}.txt", 'w') as f:
        f.write(table)


for i in range(2,21):
    generateTables(i)
```

```python
generateTables(20)
```
